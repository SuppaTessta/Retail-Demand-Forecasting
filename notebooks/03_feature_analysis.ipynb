```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load feature-engineered data
train = pd.read_csv('../data/processed/train_features.csv', parse_dates=['Date'], dtype={'StateHoliday': str})

print("="*70)
print("FEATURE ANALYSIS")
print("="*70)

print(f"\nDataset shape: {train.shape}")
print(f"Total features: {len(train.columns)}")

# Select numerical features only
numerical_cols = train.select_dtypes(include=['float64', 'int64']).columns.tolist()
print(f"\nNumerical features: {len(numerical_cols)}")

# Calculate correlation with Sales
if 'Sales' in train.columns:
    correlations = train[numerical_cols].corr()['Sales'].sort_values(ascending=False)
    
    print("\n📊 TOP 15 FEATURES CORRELATED WITH SALES:")
    print("="*70)
    for i, (feature, corr) in enumerate(correlations.head(16).items(), 1):
        if feature != 'Sales':
            print(f"  {i:2d}. {feature:40s} : {corr:6.4f}")
    
    # Visualization
    plt.figure(figsize=(12, 8))
    top_features = correlations.head(16).drop('Sales')
    top_features.plot(kind='barh', color='steelblue')
    plt.title('Top 15 Features Correlated with Sales', fontsize=14, fontweight='bold')
    plt.xlabel('Correlation with Sales')
    plt.ylabel('Feature')
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

print("\n✅ Feature analysis complete!")
```
