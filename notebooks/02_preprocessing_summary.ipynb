```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load preprocessed data
train = pd.read_csv('../data/processed/train_preprocessed.csv', parse_dates=['Date'])

print("="*70)
print("PREPROCESSING SUMMARY")
print("="*70)

print("\n📊 DATASET OVERVIEW")
print(f"  Total records: {len(train):,}")
print(f"  Date range: {train['Date'].min()} to {train['Date'].max()}")
print(f"  Total stores: {train['Store'].nunique()}")
print(f"  Total columns: {len(train.columns)}")

print("\n📋 COLUMN LIST:")
for i, col in enumerate(train.columns, 1):
    print(f"  {i:2d}. {col}")

print("\n💰 SALES STATISTICS")
print(train['Sales'].describe())

# Visualization: Sales distribution after preprocessing
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].hist(train['Sales'], bins=50, edgecolor='black', alpha=0.7, color='steelblue')
axes[0].set_title('Sales Distribution (After Preprocessing)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Sales')
axes[0].set_ylabel('Frequency')
axes[0].grid(alpha=0.3)

daily_sales = train.groupby('Date')['Sales'].sum()
axes[1].plot(daily_sales.index, daily_sales.values, linewidth=0.7, color='forestgreen')
axes[1].set_title('Total Daily Sales Over Time', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Date')
axes[1].set_ylabel('Total Sales')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✅ Data is clean and ready for feature engineering!")
```
